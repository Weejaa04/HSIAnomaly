```python
# Standard library
import os
import random

# Data processing
import numpy as np
import pandas as pd

# Visualization
import matplotlib.pyplot as plt
from PIL import Image

# Machine Learning
from sklearn.decomposition import PCA
from sklearn.metrics import (
    precision_recall_curve,
    average_precision_score,
    roc_curve,
    roc_auc_score
)

# Hyperspectral imaging
import spectral.io.envi as envi
from spectral import open_image

# Set random seed for reproducibility
SEED = 67
random.seed(SEED)
np.random.seed(SEED)

print(f"Random seed: {SEED}")
```

```python
# Configuration
DATA_DIR = "/home/alfi/projects/food-anomaly/AnomalyonFood/Dataset/Almond/Train"
TEST_DIR = "/home/alfi/projects/food-anomaly/AnomalyonFood/Dataset/Almond/Test"
N_COMPONENTS = 20  # Number of PCA components
DECISION_PERCENTILE = 99

print(f"Using SEED={SEED} for reproducibility")
print(f"PCA components: {N_COMPONENTS}")

def load_and_calibrate(data_dir):
    """
    Load hyperspectral data from data.hdr file.
    """
    # Locate data.hdr file
    data_file = None
    
    for f in os.listdir(data_dir):
        if f.endswith('.hdr') and "data" in f.lower():
            data_file = os.path.join(data_dir, f)
            break
    
    if not data_file:
        raise FileNotFoundError(f"data.hdr not found in {data_dir}")
    
    print(f"Loading data from {data_dir}:")
    print(f"  - Data: {os.path.basename(data_file)}")
    
    # Load the hyperspectral cube
    data = open_image(data_file).load().astype(np.float32)
    
    print(f"  - Data shape: {data.shape}")
    print(f"  - Data range: [{data.min():.4f}, {data.max():.4f}]")
    
    # Simple normalization to [0, 1] range
    data_min = data.min()
    data_max = data.max()
    if data_max > data_min:
        data = (data - data_min) / (data_max - data_min)
    
    print(f"  - Normalized range: [{data.min():.4f}, {data.max():.4f}]")
    
    return data

hsi_data = load_and_calibrate(DATA_DIR)
print(f"Data shape: {hsi_data.shape}")
```

# Unsupervised

```python
# Train PCA on normal data
print("\n" + "="*60)
print("Training PCA on Normal Data")
print("="*60 + "\n")

h, w, b = hsi_data.shape

# Reshape to (pixels, bands)
pixels_train = hsi_data.reshape(-1, b)
print(f"Training data shape: {pixels_train.shape}")

# Fit PCA
pca = PCA(n_components=N_COMPONENTS, random_state=SEED)
print(f"Fitting PCA with {N_COMPONENTS} components...")
pca.fit(pixels_train)

# Calculate explained variance
explained_var = np.sum(pca.explained_variance_ratio_)
print(f"\nExplained variance: {explained_var:.4f} ({explained_var*100:.2f}%)")

# Transform training data
pixels_train_pca = pca.transform(pixels_train)
print(f"PCA-transformed shape: {pixels_train_pca.shape}")

# Calculate mean of normal data in PCA space
mean_normal = np.mean(pixels_train_pca, axis=0)
print(f"Mean normal vector shape: {mean_normal.shape}")

# Compute distances for decision threshold
distances = np.linalg.norm(pixels_train_pca - mean_normal, axis=1)
decision_threshold = np.percentile(distances, DECISION_PERCENTILE)
print(f"\nDistance statistics (training normal data):")
print(f"  Mean: {distances.mean():.4f}")
print(f"  Std: {distances.std():.4f}")
print(f"  Decision threshold ({DECISION_PERCENTILE}th percentile): {decision_threshold:.4f}")

# Visualize explained variance
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.bar(range(1, N_COMPONENTS+1), pca.explained_variance_ratio_)
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('PCA Explained Variance per Component')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(range(1, N_COMPONENTS+1), np.cumsum(pca.explained_variance_ratio_), marker='o')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title(f'Cumulative Explained Variance (Total: {explained_var:.2%})')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✅ PCA model trained successfully!")
```

```python
print("Inference & Visualization...")
hsi_data_test = load_and_calibrate(TEST_DIR)
print(f"Test Data shape: {hsi_data_test.shape}")

h_test, w_test, b_test = hsi_data_test.shape

# Reshape test data to (pixels, bands)
pixels_test = hsi_data_test.reshape(-1, b_test)
print(f"Test data reshaped: {pixels_test.shape}")

# Transform to PCA space
print("Transforming test data to PCA space...")
pixels_test_pca = pca.transform(pixels_test)

# Calculate Euclidean distance from normal mean (trained on normal training data)
print("Calculating distance from normal mean...")
raw_distances = np.linalg.norm(pixels_test_pca - mean_normal, axis=1)

# Keep raw distances for metrics (ROC-AUC, PR-AUC)
anomaly_scores_flat = raw_distances

# NOTE: Rank normalization is applied ONLY for visualization.
# ROC-AUC / PR-AUC depend only on score ordering, so metrics are unaffected.
from scipy.stats import rankdata
rank_anomaly_scores = rankdata(raw_distances, method="average") / len(raw_distances)

# Reshape rank scores to image for visualization
anomaly_map = rank_anomaly_scores.reshape(h_test, w_test)
final_error_map = anomaly_map  # already in [0, 1]

# Binary decision map
decision_map = (raw_distances > decision_threshold).reshape(h_test, w_test)

# RGB image for visualization
rgb_bands = [100, 60, 20] if b_test > 100 else [max(b_test - 1, 0), max(b_test // 2, 0), max(b_test // 3, 0)]
rgb_img = hsi_data_test[:, :, rgb_bands]
rgb_img = (rgb_img - rgb_img.min()) / (rgb_img.max() - rgb_img.min() + 1e-8)

# Optional label-based metrics if label.npy exists
label_path = os.path.join(TEST_DIR, "label.npy")
labels_available = os.path.exists(label_path)
if labels_available:
    label_test = np.load(label_path)
    
    # Check and resize label if dimensions don't match
    if label_test.shape != (h_test, w_test):
        print(f"Warning: Label shape {label_test.shape} doesn't match HSI shape ({h_test}, {w_test})")
        from scipy.ndimage import zoom
        scale_h = h_test / label_test.shape[0]
        scale_w = w_test / label_test.shape[1]
        label_test = zoom(label_test, (scale_h, scale_w), order=0)
        print(f"Resized label to {label_test.shape}")
    
    label_flat = label_test.flatten()
    label_binary = (label_flat > label_flat.mean()).astype(int)
    aupr_score = average_precision_score(label_binary, anomaly_scores_flat)
    roc_auc = roc_auc_score(label_binary, anomaly_scores_flat)
    precision, recall, _ = precision_recall_curve(label_binary, anomaly_scores_flat)
    fpr, tpr, _ = roc_curve(label_binary, anomaly_scores_flat)
else:
    aupr_score = None
    roc_auc = None
    precision = recall = fpr = tpr = None

fig = plt.figure(figsize=(20, 12))
gs = fig.add_gridspec(3, 4, hspace=0.3, wspace=0.3)

# Row 1: Calibration Image, False-Color HSI, Predicted Anomaly Map, Overlay
ax1 = fig.add_subplot(gs[0, 0])
calib_path = os.path.join(TEST_DIR, "image_calibration.png")
if os.path.exists(calib_path):
    calib_img = Image.open(calib_path)
    ax1.imshow(calib_img)
    ax1.set_title('Calibration Image', fontsize=12, fontweight='bold')
else:
    ax1.text(0.5, 0.5, 'Calibration Image\nNot Found', ha='center', va='center', fontsize=12)
    ax1.set_title('Calibration Image', fontsize=12, fontweight='bold')
ax1.axis('off')

ax2 = fig.add_subplot(gs[0, 1])
if labels_available:
    # Convert to binary: minimum value = 0, others = 1
    label_test_binary = (label_test > label_test.min()).astype(int)
    
    # Create custom colormap: green for 0, red for 1
    from matplotlib.colors import ListedColormap
    cmap_binary = ListedColormap(['green', 'red'])
    
    im2 = ax2.imshow(label_test_binary, cmap=cmap_binary)
    ax2.set_title('Ground Truth Labels (Binary)', fontsize=12, fontweight='bold')
    ax2.axis('off')
    plt.colorbar(im2, ax=ax2, fraction=0.046, pad=0.04)
else:
    ax2.text(0.5, 0.5, 'Labels Not Available', ha='center', va='center', fontsize=12)
    ax2.set_title('Ground Truth Labels', fontsize=12, fontweight='bold')
    ax2.axis('off')

ax3 = fig.add_subplot(gs[0, 2])
im3 = ax3.imshow(final_error_map, cmap='jet')
ax3.set_title('Predicted Anomaly Map', fontsize=12, fontweight='bold')
ax3.axis('off')
plt.colorbar(im3, ax=ax3, fraction=0.046, pad=0.04)

ax4 = fig.add_subplot(gs[0, 3])
ax4 = fig.add_subplot(gs[0, 3])
ax4.imshow(rgb_img)
ax4.imshow(decision_map, cmap='Reds', alpha=0.6)
ax4.set_title(f'Decision Map (>{DECISION_PERCENTILE}th percentile)', fontsize=12, fontweight='bold')
ax4.axis('off')

# Row 2: Precision-Recall Curve and ROC Curve
ax5 = fig.add_subplot(gs[1, 0:2])
if labels_available:
    ax5.plot(recall, precision, color='darkorange', lw=2, label=f'PR curve (AUPR = {aupr_score:.4f})')
    baseline = label_binary.sum() / len(label_binary)
    ax5.plot([0, 1], [baseline, baseline], color='navy', lw=2, linestyle='--', label=f'Random classifier (AP = {baseline:.4f})')
    ax5.set_xlim([0.0, 1.0])
    ax5.set_ylim([0.0, 1.05])
    ax5.set_xlabel('Recall', fontsize=11)
    ax5.set_ylabel('Precision', fontsize=11)
    ax5.set_title('Precision-Recall Curve', fontsize=12, fontweight='bold')
    ax5.legend(loc='best')
    ax5.grid(True, alpha=0.3)
else:
    ax5.text(0.5, 0.5, 'No labels found\nPR curve unavailable', ha='center', va='center', fontsize=11)
    ax5.set_title('Precision-Recall Curve', fontsize=12, fontweight='bold')
    ax5.axis('off')

ax6 = fig.add_subplot(gs[1, 2:4])
if labels_available:
    ax6.plot(fpr, tpr, color='darkblue', lw=2, label=f'ROC curve (AUC = {roc_auc:.4f})')
    ax6.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random classifier (AUC = 0.5)')
    ax6.set_xlim([0.0, 1.0])
    ax6.set_ylim([0.0, 1.05])
    ax6.set_xlabel('False Positive Rate', fontsize=11)
    ax6.set_ylabel('True Positive Rate', fontsize=11)
    ax6.set_title('ROC Curve', fontsize=12, fontweight='bold')
    ax6.legend(loc='lower right')
    ax6.grid(True, alpha=0.3)
else:
    ax6.text(0.5, 0.5, 'No labels found\nROC curve unavailable', ha='center', va='center', fontsize=11)
    ax6.set_title('ROC Curve', fontsize=12, fontweight='bold')
    ax6.axis('off')

# Row 3: Score Distribution and Statistics
ax7 = fig.add_subplot(gs[2, 0:2])
if labels_available:
    ax7.hist(anomaly_scores_flat[label_binary == 0], bins=50, alpha=0.5, label='Normal', color='green', density=True)
    ax7.hist(anomaly_scores_flat[label_binary == 1], bins=50, alpha=0.5, label='Anomaly', color='red', density=True)
    ax7.set_xlabel('Anomaly Score', fontsize=11)
    ax7.set_ylabel('Density', fontsize=11)
    ax7.set_title('Score Distribution', fontsize=12, fontweight='bold')
    ax7.legend()
    ax7.grid(True, alpha=0.3)
else:
    ax7.hist(anomaly_scores_flat, bins=50, color='crimson', alpha=0.7, density=True)
    ax7.set_xlabel('Anomaly Score', fontsize=11)
    ax7.set_ylabel('Density', fontsize=11)
    ax7.set_title('Score Distribution', fontsize=12, fontweight='bold')
    ax7.grid(True, alpha=0.3)

ax8 = fig.add_subplot(gs[2, 2:4])
if labels_available:
    stats_text = f"""MODEL PERFORMANCE
{'='*30}

AUPR Score:    {aupr_score:.4f}
ROC-AUC Score: {roc_auc:.4f}

Method: PCA + Distance
PCA Components: {N_COMPONENTS}
Explained Variance: {explained_var:.2%}

Dataset: Test Split
Total Pixels: {len(anomaly_scores_flat):,}
Anomalies: {label_binary.sum():,} ({100*label_binary.sum()/len(label_binary):.1f}%)
Normal: {(1-label_binary).sum():,} ({100*(1-label_binary).sum()/len(label_binary):.1f}%)

Score Statistics:
Mean: {anomaly_scores_flat.mean():.4f}
Std:  {anomaly_scores_flat.std():.4f}
Min:  {anomaly_scores_flat.min():.4f}
Max:  {anomaly_scores_flat.max():.4f}
"""
else:
    stats_text = f"""MODEL PERFORMANCE
{'='*30}

AUPR Score:    unavailable (no labels)
ROC-AUC Score: unavailable (no labels)

Method: PCA + Distance
PCA Components: {N_COMPONENTS}
Explained Variance: {explained_var:.2%}

Dataset: Test Split
Total Pixels: {len(anomaly_scores_flat):,}

Score Statistics:
Mean: {anomaly_scores_flat.mean():.4f}
Std:  {anomaly_scores_flat.std():.4f}
Min:  {anomaly_scores_flat.min():.4f}
Max:  {anomaly_scores_flat.max():.4f}
"""
ax8.text(0.1, 0.5, stats_text, fontsize=10, family='monospace', verticalalignment='center', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))
ax8.axis('off')

plt.suptitle(f'Anomaly Detection (PCA + Distance)' + (f' (AUPR: {aupr_score:.4f} | ROC-AUC: {roc_auc:.4f})' if labels_available else ''), 
             fontsize=14, fontweight='bold', y=0.98)
plt.show()

print(f"\n✅ Evaluation complete!")
if labels_available:
    print(f"   AUPR Score: {aupr_score:.4f}")
    print(f"   ROC-AUC Score: {roc_auc:.4f}")
```

```python
print("\n" + "="*60)
print("MAHALANOBIS DISTANCE VARIANT")
print("="*60 + "\n")

from scipy.spatial.distance import mahalanobis

print("Computing covariance matrix from training data...")
# Calculate covariance matrix of PCA-transformed training data
cov_matrix = np.cov(pixels_train_pca.T)
print(f"Covariance matrix shape: {cov_matrix.shape}")

# Compute inverse of covariance matrix
cov_inv = np.linalg.inv(cov_matrix)
print("Covariance matrix inverted")

# Calculate Mahalanobis distance from normal mean
print("Calculating Mahalanobis distance from normal mean...")
raw_distances_maha = np.array([
    mahalanobis(p, mean_normal, cov_inv) for p in pixels_test_pca
])

# Keep raw distances for metrics
anomaly_scores_flat_maha = raw_distances_maha

# NOTE: Rank normalization is applied ONLY for visualization.
from scipy.stats import rankdata
rank_anomaly_scores_maha = rankdata(raw_distances_maha, method="average") / len(raw_distances_maha)

# Reshape rank scores to image for visualization
anomaly_map_maha = rank_anomaly_scores_maha.reshape(h_test, w_test)
final_error_map_maha = anomaly_map_maha  # already in [0, 1]

# Binary decision map using Mahalanobis distance threshold
decision_threshold_maha = np.percentile(raw_distances_maha, DECISION_PERCENTILE)
decision_map_maha = (raw_distances_maha > decision_threshold_maha).reshape(h_test, w_test)

# Optional label-based metrics if label.npy exists
if labels_available:
    aupr_score_maha = average_precision_score(label_binary, anomaly_scores_flat_maha)
    roc_auc_maha = roc_auc_score(label_binary, anomaly_scores_flat_maha)
    precision_maha, recall_maha, _ = precision_recall_curve(label_binary, anomaly_scores_flat_maha)
    fpr_maha, tpr_maha, _ = roc_curve(label_binary, anomaly_scores_flat_maha)
else:
    aupr_score_maha = None
    roc_auc_maha = None
    precision_maha = recall_maha = fpr_maha = tpr_maha = None

fig_maha = plt.figure(figsize=(20, 12))
gs_maha = fig_maha.add_gridspec(3, 4, hspace=0.3, wspace=0.3)

# Row 1: Calibration Image, False-Color HSI, Predicted Anomaly Map, Overlay
ax1_maha = fig_maha.add_subplot(gs_maha[0, 0])
if os.path.exists(calib_path):
    calib_img = Image.open(calib_path)
    ax1_maha.imshow(calib_img)
    ax1_maha.set_title('Calibration Image', fontsize=12, fontweight='bold')
else:
    ax1_maha.text(0.5, 0.5, 'Calibration Image\nNot Found', ha='center', va='center', fontsize=12)
    ax1_maha.set_title('Calibration Image', fontsize=12, fontweight='bold')
ax1_maha.axis('off')

ax2_maha = fig_maha.add_subplot(gs_maha[0, 1])
if labels_available:
    label_test_binary = (label_test > label_test.min()).astype(int)
    from matplotlib.colors import ListedColormap
    cmap_binary = ListedColormap(['green', 'red'])
    im2_maha = ax2_maha.imshow(label_test_binary, cmap=cmap_binary)
    ax2_maha.set_title('Ground Truth Labels (Binary)', fontsize=12, fontweight='bold')
    ax2_maha.axis('off')
    plt.colorbar(im2_maha, ax=ax2_maha, fraction=0.046, pad=0.04)
else:
    ax2_maha.text(0.5, 0.5, 'Labels Not Available', ha='center', va='center', fontsize=12)
    ax2_maha.set_title('Ground Truth Labels', fontsize=12, fontweight='bold')
    ax2_maha.axis('off')

ax3_maha = fig_maha.add_subplot(gs_maha[0, 2])
im3_maha = ax3_maha.imshow(final_error_map_maha, cmap='jet')
ax3_maha.set_title('Predicted Anomaly Map (Mahalanobis)', fontsize=12, fontweight='bold')
ax3_maha.axis('off')
plt.colorbar(im3_maha, ax=ax3_maha, fraction=0.046, pad=0.04)

ax4_maha = fig_maha.add_subplot(gs_maha[0, 3])
ax4_maha.imshow(rgb_img)
ax4_maha.imshow(decision_map_maha, cmap='Reds', alpha=0.6)
ax4_maha.set_title(f'Decision Map (>{DECISION_PERCENTILE}th percentile)', fontsize=12, fontweight='bold')
ax4_maha.axis('off')

# Row 2: Precision-Recall Curve and ROC Curve
ax5_maha = fig_maha.add_subplot(gs_maha[1, 0:2])
if labels_available:
    ax5_maha.plot(recall_maha, precision_maha, color='darkorange', lw=2, label=f'PR curve (AUPR = {aupr_score_maha:.4f})')
    baseline = label_binary.sum() / len(label_binary)
    ax5_maha.plot([0, 1], [baseline, baseline], color='navy', lw=2, linestyle='--', label=f'Random classifier (AP = {baseline:.4f})')
    ax5_maha.set_xlim([0.0, 1.0])
    ax5_maha.set_ylim([0.0, 1.05])
    ax5_maha.set_xlabel('Recall', fontsize=11)
    ax5_maha.set_ylabel('Precision', fontsize=11)
    ax5_maha.set_title('Precision-Recall Curve', fontsize=12, fontweight='bold')
    ax5_maha.legend(loc='best')
    ax5_maha.grid(True, alpha=0.3)
else:
    ax5_maha.text(0.5, 0.5, 'No labels found\nPR curve unavailable', ha='center', va='center', fontsize=11)
    ax5_maha.set_title('Precision-Recall Curve', fontsize=12, fontweight='bold')
    ax5_maha.axis('off')

ax6_maha = fig_maha.add_subplot(gs_maha[1, 2:4])
if labels_available:
    ax6_maha.plot(fpr_maha, tpr_maha, color='darkblue', lw=2, label=f'ROC curve (AUC = {roc_auc_maha:.4f})')
    ax6_maha.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random classifier (AUC = 0.5)')
    ax6_maha.set_xlim([0.0, 1.0])
    ax6_maha.set_ylim([0.0, 1.05])
    ax6_maha.set_xlabel('False Positive Rate', fontsize=11)
    ax6_maha.set_ylabel('True Positive Rate', fontsize=11)
    ax6_maha.set_title('ROC Curve', fontsize=12, fontweight='bold')
    ax6_maha.legend(loc='lower right')
    ax6_maha.grid(True, alpha=0.3)
else:
    ax6_maha.text(0.5, 0.5, 'No labels found\nROC curve unavailable', ha='center', va='center', fontsize=11)
    ax6_maha.set_title('ROC Curve', fontsize=12, fontweight='bold')
    ax6_maha.axis('off')

# Row 3: Score Distribution and Statistics
ax7_maha = fig_maha.add_subplot(gs_maha[2, 0:2])
if labels_available:
    ax7_maha.hist(anomaly_scores_flat_maha[label_binary == 0], bins=50, alpha=0.5, label='Normal', color='green', density=True)
    ax7_maha.hist(anomaly_scores_flat_maha[label_binary == 1], bins=50, alpha=0.5, label='Anomaly', color='red', density=True)
    ax7_maha.set_xlabel('Anomaly Score', fontsize=11)
    ax7_maha.set_ylabel('Density', fontsize=11)
    ax7_maha.set_title('Score Distribution', fontsize=12, fontweight='bold')
    ax7_maha.legend()
    ax7_maha.grid(True, alpha=0.3)
else:
    ax7_maha.hist(anomaly_scores_flat_maha, bins=50, color='crimson', alpha=0.7, density=True)
    ax7_maha.set_xlabel('Anomaly Score', fontsize=11)
    ax7_maha.set_ylabel('Density', fontsize=11)
    ax7_maha.set_title('Score Distribution', fontsize=12, fontweight='bold')
    ax7_maha.grid(True, alpha=0.3)

ax8_maha = fig_maha.add_subplot(gs_maha[2, 2:4])
if labels_available:
    stats_text_maha = f"""MODEL PERFORMANCE
{'='*30}

AUPR Score:    {aupr_score_maha:.4f}
ROC-AUC Score: {roc_auc_maha:.4f}

Method: PCA + Mahalanobis Distance
PCA Components: {N_COMPONENTS}
Explained Variance: {explained_var:.2%}

Dataset: Test Split
Total Pixels: {len(anomaly_scores_flat_maha):,}
Anomalies: {label_binary.sum():,} ({100*label_binary.sum()/len(label_binary):.1f}%)
Normal: {(1-label_binary).sum():,} ({100*(1-label_binary).sum()/len(label_binary):.1f}%)

Score Statistics:
Mean: {anomaly_scores_flat_maha.mean():.4f}
Std:  {anomaly_scores_flat_maha.std():.4f}
Min:  {anomaly_scores_flat_maha.min():.4f}
Max:  {anomaly_scores_flat_maha.max():.4f}
"""
else:
    stats_text_maha = f"""MODEL PERFORMANCE
{'='*30}

AUPR Score:    unavailable (no labels)
ROC-AUC Score: unavailable (no labels)

Method: PCA + Mahalanobis Distance
PCA Components: {N_COMPONENTS}
Explained Variance: {explained_var:.2%}

Dataset: Test Split
Total Pixels: {len(anomaly_scores_flat_maha):,}

Score Statistics:
Mean: {anomaly_scores_flat_maha.mean():.4f}
Std:  {anomaly_scores_flat_maha.std():.4f}
Min:  {anomaly_scores_flat_maha.min():.4f}
Max:  {anomaly_scores_flat_maha.max():.4f}
"""
ax8_maha.text(0.1, 0.5, stats_text_maha, fontsize=10, family='monospace', verticalalignment='center', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))
ax8_maha.axis('off')

plt.suptitle(f'Anomaly Detection (PCA + Mahalanobis Distance)' + (f' (AUPR: {aupr_score_maha:.4f} | ROC-AUC: {roc_auc_maha:.4f})' if labels_available else ''), 
             fontsize=14, fontweight='bold', y=0.98)
plt.show()

print(f"\n✅ Mahalanobis Distance evaluation complete!")
if labels_available:
    print(f"   AUPR Score: {aupr_score_maha:.4f}")
    print(f"   ROC-AUC Score: {roc_auc_maha:.4f}")
```
